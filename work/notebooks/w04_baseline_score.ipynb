## My Rule

A page should be reviewed if it is stale, receives meaningful search visibility, and shows signs of underperforming in search results. I use content freshness, impressions, CTR, and average position as transparent signals that can be understood by both technical and non-technical stakeholders.

### Reason Code

- `STALE_LOW_CTR`
- **Action Label:** Refresh Content

```python
# Create simple rule signals

df["stale"] = (df["days_since_last_update"] >= 180).astype(int)
df["visible"] = (df["impressions_90d"] >= 500).astype(int)
df["low_ctr"] = (df["ctr"] < 2).astype(int)
df["poor_position"] = (df["avg_position"] > 10).astype(int)

print(df[["stale","visible","low_ctr","poor_position"]].head())
```

## Baseline Rule

The baseline assigns higher scores to pages that are stale, receive enough impressions to matter, have low CTR, and rank below the first page of search results. Every flagged page receives one transparent reason code and one action label.

This rule is intentionally simple so it can serve as a baseline that a machine learning model should outperform in Week 5.

```python
# Transparent baseline score

df["baseline_score"] = (
    df["stale"] * 40 +
    df["visible"] * 20 +
    df["low_ctr"] * 20 +
    df["poor_position"] * 20
)

df["reason_code"] = "STALE_LOW_CTR"
df["action"] = "Refresh Content"

ranked = df.sort_values("baseline_score", ascending=False)

from pathlib import Path

output_dir = Path("work/outputs")
output_dir.mkdir(parents=True, exist_ok=True)

ranked.to_csv(output_dir / "baseline_action_score.csv", index=False)

print(ranked[[
    "baseline_score",
    "reason_code",
    "action",
    "days_since_last_update",
    "ctr",
    "avg_position",
    "impressions_90d"
]].head(10))

print("\nCSV written to:", output_dir / "baseline_action_score.csv")
```

## Weak Picks

The baseline is intentionally simple and may rank some pages higher than they deserve. Weak picks may occur when:

- A page is seasonal and naturally loses traffic.
- CTR is unreliable because impressions are very low.
- A page is already scheduled for an update.
- Search demand has shifted rather than content quality declining.
- Business priorities differ from search performance.

These cases require manual review before action is taken.
