## Unit of Analysis

Each row in the dataset represents one webpage.

Each webpage contains features such as:

- impressions
- clicks
- CTR
- average position
- content age
- days since last update
- engagement

The model learns patterns from these features.

## Success Metric

The main evaluation metric is **Precision@K**.

Editors usually review only a limited number of pages, so Precision@K measures how many of the highest-ranked pages actually need attention.
